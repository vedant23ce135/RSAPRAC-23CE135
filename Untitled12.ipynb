```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np


sel_m_base = metric_frame.by_group['selection_rate']['Male']
sel_f_base = metric_frame.by_group['selection_rate']['Female']

sel_m_dp = metric_frame_dp.by_group['selection_rate']['Male']
sel_f_dp = metric_frame_dp.by_group['selection_rate']['Female']

sel_m_eo = metric_frame_eo.by_group['selection_rate']['Male']
sel_f_eo = metric_frame_eo.by_group['selection_rate']['Female']


comparison_data = {
    'MODEL': ['Baseline', 'TO: demographic_par.', 'TO: equalized_odds'],
    'ACC': [metric_frame.overall['accuracy'], metric_frame_dp.overall['accuracy'], metric_frame_eo.overall['accuracy']],
    'SEL_M': [sel_m_base, sel_m_dp, sel_m_eo],
    'SEL_F': [sel_f_base, sel_f_dp, sel_f_eo],
    'DP_DIFF': [dp_diff, dp_diff_dp, dp_diff_eo],
    'EO_DIFF': [eo_diff, eo_diff_dp, eo_diff_eo]
}

comparison_df = pd.DataFrame(comparison_data)

print("Expected Output — Part B: Mitigation with ThresholdOptimizer")
print("MODEL                  ACC    SEL_M  SEL_F  DP_DIFF  EO_DIFF")
print("--------------------------------------------------------------")
for idx, row in comparison_df.iterrows():
    print(f"{row['MODEL']:<22} {row['ACC']:.2f}   {row['SEL_M']:.2f}   {row['SEL_F']:.2f}   {row['DP_DIFF']:.2f}     {row['EO_DIFF']:.2f}")
print("--------------------------------------------------------------")


labels = ['Baseline', 'TO: demographic_par.', 'TO: equalized_odds']
male_rates = [sel_m_base, sel_m_dp, sel_m_eo]
female_rates = [sel_f_base, sel_f_dp, sel_f_eo]

x = np.arange(len(labels))
width = 0.35

fig, ax = plt.subplots(figsize=(8, 6))
rects1 = ax.bar(x - width/2, male_rates, width, label='Male', color='skyblue')
rects2 = ax.bar(x + width/2, female_rates, width, label='Female', color='salmon')

ax.set_ylabel('Selection Rate')
ax.set_title('Selection Rate by Sex across Models')
ax.set_xticks(x)
ax.set_xticklabels(labels)
ax.legend()
ax.set_ylim(0, 0.35)

plt.tight_layout()
plt.savefig('/content/selection_rate_by_sex.png')
plt.show()
print("Plot saved : selection_rate_by_sex.png")
```
